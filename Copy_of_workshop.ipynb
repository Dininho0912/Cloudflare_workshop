<details><summary>💡 Hint 2: How do I read my secret and call the API?</summary>

- Read the token with `userdata.get("RADAR_TOKEN")`.

</details>

<details><summary>💡 Hint 3: I get a big JSON, where's my answer?</summary>

Look inside `result`, then `summary_0`.
</details>

````python
# 1. Read your token from secrets
token = userdata.get("RADAR_TOKEN")

# 2. Call the Radar API endpoint for Device Type summary
url = "https://api.cloudflare.com/client/v4/radar/http/summary/DEVICE_TYPE"
headers = {
    "Authorization": f"Bearer {token}"
}
params = {
    "dateRange": "7d"
}

response = requests.get(url, headers=headers, params=params)
response.raise_for_status() # Make sure the request succeeded

# 3. Extract the summary data from the JSON result
data = response.json()["result"]["summary_0"]
print(data)
````

```python
TOKEN = userdata.get("RADAR_TOKEN")

def radar(path, **params):
    """Call the Radar API and return the 'result' part."""
    r = requests.get(f"https://api.cloudflare.com/client/v4/radar/{path}",
                     params=params,
                     headers={"Authorization": f"Bearer {TOKEN}"})
    r.raise_for_status()
    return r.json()["result"]
```

</details>

````python
TOKEN = userdata.get("RADAR_TOKEN")

def radar(path, **params):
  # Changed "path = params" to "params = params"
  r = requests.get(f"https://api.cloudflare.com/client/v4/radar/{path}",
                   params = params,
                   headers={"Authorization": f"Bearer {TOKEN}"})

  r.raise_for_status()
  return r.json()["result"]
````

### **Challenge 2: The day Portugal went dark 🔌**

Something big happened to Portugal's Internet traffic in **April 2025**.

**Question:** can you find **what day** and **what time** it happened, just by looking at the data?

**Goal:** plot Portugal's **human** Internet traffic, **hour by hour**, for the **whole month of April 2025**.

**You got it if** you see a line chart with daily waves and **one day that looks very different**.

<details><summary>💡 Hint 1: Which data?</summary>

You need traffic **over time**, not a summary. Look in the docs for an HTTP **timeseries** endpoint.

</details>

<details><summary>💡 Hint 2: How do I choose the dates?</summary>

Instead of `dateRange`, use `dateStart` and `dateEnd`.
They use this format: `2025-04-01T00:00:00Z`

To get one point per hour, look for a parameter called `aggInterval`.

</details>

<details><summary>💡 Hint 3: Where are the numbers?</summary>

Inside `result` → `serie_0` you'll find two lists: `timestamps` and `values`.
`pd.to_datetime(...)` turns the timestamps into real dates for your chart.

</details>

<details><summary>💡 Hint 4: The times look wrong!</summary>

The API uses **UTC** time. Try `.tz_convert("Europe/Lisbon")` on your index. 🕐

</details>

````python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Call the timeseries API for Portugal in April 2025
# Endpoint: http/timeseries
result = radar(
    "http/timeseries",
    dateStart="2025-04-01T00:00:00Z",
    dateEnd="2025-05-01T00:00:00Z",
    aggInterval="1h",
    location="PT",
    botClass="LIKELY_HUMAN"
)

# 2. Extract timestamps and values
# The series is usually in result["serie_0"]
timestamps = result["serie_0"]["timestamps"]
values = result["serie_0"]["values"]

# 3. Build a DataFrame
df_outage = pd.DataFrame({
    "traffic": values
}, index=pd.to_datetime(timestamps))

# Convert timezone to Europe/Lisbon as suggested
df_outage.index = df_outage.index.tz_convert("Europe/Lisbon")

# Convert values to float
df_outage["traffic"] = df_outage["traffic"].astype(float)

# 4. Plot the graph
plt.figure(figsize=(15, 6))
plt.plot(df_outage.index, df_outage["traffic"], color="#ff7f0e")
plt.title("Portugal Human Traffic - April 2025 (Hour-by-Hour)")
plt.xlabel("Date")
plt.ylabel("Traffic Index")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
````

🔍 **Explore:**
- At what **time** did the drop start?

````python
# Find the lowest traffic point in April 2025
lowest_point = df_outage["traffic"].idxmin()
lowest_value = df_outage["traffic"].min()

print(f"The lowest traffic was recorded at: {lowest_point}")
print(f"Traffic index at lowest point: {lowest_value:.4f}")

# Let's show the data around that specific day to pinpoint when the drop started
# We filter for April 28, 2025
df_april_28 = df_outage.loc["2025-04-28"]
display(df_april_28)
````

🔍 **Explore:**
- How long did it take to come back to normal?

````python
# Let's zoom in on the period between April 27th and April 30th to see the recovery curve
plt.figure(figsize=(12, 5))
plt.plot(df_outage.loc["2025-04-27":"2025-04-30"].index, df_outage.loc["2025-04-27":"2025-04-30"]["traffic"], marker='o', color='red')
plt.title("Zooming In: The Outage and Recovery (April 27 - April 30)")
plt.xlabel("Time")
plt.ylabel("Traffic Index")
plt.grid(True)
plt.show()

print("By looking at the hourly markers on the plot, we can calculate how many hours the traffic index remained abnormally low.")
````
